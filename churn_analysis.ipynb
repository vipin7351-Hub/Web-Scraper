```python
# import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
import sqlite3
```

```python
# 1. import database / data 
conn = sqlite3.connect('customer_churn.db')
sql_query = """
        SELECT name
        FROM sqlite_master
        WHERE type='table'
   """
tables = pd.read_sql(sql_query, conn)

# create dataFrame for each table
for table_name in tables['name']:
    df = pd.read_sql(f"SELECT * FROM {table_name}", conn)
    globals()[f"df_{table_name}"] = df
    print(f"Create dataframe: df_{table_name}")

conn.close()
```

```python
#find columns name
conn = sqlite3.connect('customer_churn.db')

for table_name in tables['name']:
    print(f"\nTable Name: {table_name}")
    #Get column information
    columns_query  = f"PRAGMA table_info({table_name});" # PRAGMA query to help read the table info
    columns = pd.read_sql(columns_query, conn)
    print("columns :")
    print(columns['name'].tolist())

# Close connection 
conn.close()
```

```python
# 5 . ARPU - Avg Revenue Per User

arpu = df_merge_data['monthly_charges'].mean().round(2)
print('ARPU = ', round(arpu,2))
```

```python
# Calculate customer age 
today  = pd.Timestamp.today()
customer_age = np.where(df_merge_data['dob'].notna(),
                        ((today - df_merge_data['dob']).dt.days/365.25).astype(int),
                        'Date of Birth not found'
                        )
# if want months : 
# today = pd.Timestamp.today()

# customer_age_months = np.where(
#     df_merge_data['dob'].notna(),
#     ((today - df_merge_data['dob']).dt.days / 30.4375).astype(int),
#     'Date of Birth not found'
# )

# print("customer age in months:", customer_age_months)


print("customer age in years  :", customer_age)
```

```python
# 6. Avg customer Tenure
# Count of days users has used our service : cancellation date else current date
today  = pd.Timestamp.today()
df_merge_data['tenure_days'] = np.where(
            df_merge_data['cancellation_date'].notna(),
            (df_merge_data['cancellation_date'] - df_merge_data['subscription_start_date']).dt.days,
            (today - df_merge_data['subscription_start_date']).dt.days
)

avg_tenure = df_merge_data['tenure_days'].mean()
print('Avg Tenure (Days)= ', round(avg_tenure))
```

```python
# 7. Revenue at risk   - revenue lost from churned users
revenue_at_risk = df_merge_data.loc[df_merge_data['churn_flag']==1 , 'monthly_charges'].sum()
print("Revenue at risk(Rs 'K') =", revenue_at_risk)
```

```python
# 8. Esclation Rate  :-
esclation_rate = (df_merge_data['escalations']=="Y").mean()*100
print("Escalation Rate = ", round(esclation_rate, 2), "%")
```

```python
# 9. Avg Complaint  per User
avg_complaints = df_merge_data['complaint_count'].sum()/ df_merge_data['customerid'].nunique()
print('Avg compliants per user = ', round(avg_complaints, 2))
```

```python
# 10. Correlation Esclation vs Churn
#df_merge_data['escalations'] = np.where(df_merge_data['escalations']=='Y',1,0) This code of line run only one time if i again run so this change all the value
corr_df = df_merge_data[['escalations', 'churn_flag']].dropna()

correlation = corr_df['escalations'].corr(df_merge_data['churn_flag'])
print("Correlation between esclation vs churn is =", round(correlation, 2))
```

```python
# 11 .chure risk - create a column using Existing col 
conditions = [
    (df_merge_data['churn_score']<50),
    (df_merge_data['churn_score']>=50) & (df_merge_data['churn_score']<70),
    (df_merge_data['churn_score']>=70)
]
choices = ['low','med', 'high']
df_merge_data['churn_risk'] = np.select(conditions, choices, default='unkown')
```
